```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
dataset = pd.read_csv('Mall_Customers.csv')
```

```python
dataset
```

```python
X = dataset.iloc[:, 3:5].values
```

```python
X
```

```python
from sklearn.cluster import AffinityPropagation

clustering = AffinityPropagation(random_state=5)
y_affinity = clustering.fit_predict(X)

unsupervised = dataset.copy()
unsupervised['Cluster_group'] = y_affinity

import seaborn as sns

facet = sns.lmplot(
    data=unsupervised,
    x=unsupervised.columns[3],
    y=unsupervised.columns[4],
    hue='Cluster_group',
    fit_reg=False,
    legend=True,
    legend_out=True
)
```

```python
# Cluster label assigned to each data point
print(clustering.labels_)

# Indices of the exemplar data points
print(clustering.cluster_centers_indices_)

# Number of clusters identified
print(len(set(clustering.labels_)))
```
