```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
dataset = pd.read_csv('Mall_Customers.csv')
```

```python
dataset
```

```python
X = dataset.iloc[:, 3:5].values
```

```python
X
```

```python
from sklearn.cluster import MeanShift, estimate_bandwidth

bandwidth = estimate_bandwidth(X, quantile=0.2)

print("Bandwidth:", bandwidth)

clustering = MeanShift(bandwidth=bandwidth)
y_meanshift = clustering.fit_predict(X)

unsupervised = dataset.copy()
unsupervised['Cluster_group'] = y_meanshift

print("Number of clusters:", len(set(y_meanshift)))
unsupervised

import seaborn as sns

facet = sns.lmplot(
    data=unsupervised,
    x=unsupervised.columns[3],
    y=unsupervised.columns[4],
    hue='Cluster_group',
    fit_reg=False,
    legend=True,
    legend_out=True
)
```
