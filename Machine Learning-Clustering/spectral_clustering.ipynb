```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
dataset = pd.read_csv('Mall_Customers.csv')
```

```python
dataset
```

```python
X = dataset.iloc[:, 3:5].values
```

```python
X
```

```python
from sklearn.cluster import SpectralClustering

clustering = SpectralClustering(
    n_clusters=5,
    random_state=42,
    assign_labels='discretize'
)
y_spectral = clustering.fit_predict(X)

unsupervised = dataset.copy()
unsupervised['Cluster_group'] = y_spectral

unsupervised


import seaborn as sns

facet = sns.lmplot(
    data=unsupervised,
    x=unsupervised.columns[3],
    y=unsupervised.columns[4],
    hue='Cluster_group',
    fit_reg=False,
    legend=True,
    legend_out=True
)
```

```python
from sklearn.cluster import SpectralClustering

clustering = SpectralClustering(
    n_clusters=5,
    random_state=42,
    assign_labels='kmeans'
)
y_spectral = clustering.fit_predict(X)

unsupervised = dataset.copy()
unsupervised['Cluster_group'] = y_spectral

unsupervised


import seaborn as sns

facet = sns.lmplot(
    data=unsupervised,
    x=unsupervised.columns[3],
    y=unsupervised.columns[4],
    hue='Cluster_group',
    fit_reg=False,
    legend=True,
    legend_out=True
)
```
